```python
# Name  : Arham Farid
# Lane  : IT
# Date  : 24/09/2026


label = input("Enter text: ")
first = float(input("Enter first number: "))
second = float(input("Enter second number"))

difference = second - first
percent = (first/ second) * 100

print()
print("=" * 34)
print(f"  RECORD CHECK  -  {label}")
print("=" * 34)

print(f"  {label:<10} {first:>10.2f}")
print(f"  {'Second':<10} {second:>10.2f}")
print(f"  {'Diffrence':<10} {difference:+10.2f}")
print(f"  {'percent':<10} {percent:>10.2f}%")
print(f" Report completed successfully. ")

print("=" * 34)
```
